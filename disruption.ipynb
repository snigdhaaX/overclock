```python
def winsor_minmax(s, lo=0.01, hi=0.99):
    """Scale to 0-1 after clipping extreme tails, so outliers don't crush everyone else."""
    a, b = s.quantile(lo), s.quantile(hi)
    return ((s.clip(a, b) - a) / (b - a))

# Fulfilment: already a rate near 1 -> use directly (higher = better)
df["fulfil_score"] = df["order_fulfillment_status"].clip(0, 1)

# Cargo: ASSUMED to be a damage rate (lower = better) -> invert
df["cargo_score"] = 1 - winsor_minmax(df["cargo_condition_status"])
```

```python
SUP, CTRY = "supplier_id", "supplier_country"

SEED = 42
rng = np.random.default_rng(SEED)
START = pd.Timestamp("2023-01-01")
n = len(df)

# Supplier-level latent quality (separates suppliers of the same product)
sup_quality = pd.Series(rng.beta(5, 2, df[SUP].nunique()), index=df[SUP].unique())
q = df[SUP].map(sup_quality).to_numpy()
ctry_transit = pd.Series(rng.integers(3, 26, df[CTRY].nunique()), index=df[CTRY].unique())

# Planned timeline
order_date = START + pd.to_timedelta(rng.integers(0, 730, n), unit="D")
planned_proc = rng.integers(1, 6, n)
planned_transit = np.clip(df[CTRY].map(ctry_transit).to_numpy() + rng.integers(-2, 3, n), 2, None)
expected_dispatch = order_date + pd.to_timedelta(planned_proc, unit="D")
expected_delivery = expected_dispatch + pd.to_timedelta(planned_transit, unit="D")

# Actual timeline (lower-quality suppliers -> more delay)
dispatch_delay = rng.poisson((1 - q) * 3)
transit_delay = rng.poisson((1 - q) * 3)
early = rng.binomial(1, 0.25 * q) * rng.integers(1, 3, n)
actual_dispatch = expected_dispatch + pd.to_timedelta(dispatch_delay, unit="D")
actual_delivery = actual_dispatch + pd.to_timedelta(planned_transit + transit_delay - early, unit="D")

df["order_date"] = order_date
df["dispatch_date"] = actual_dispatch
df["expected_delivery_date"] = expected_delivery
df["actual_delivery_date"] = actual_delivery

# Keep the dataset's original deviation for reference only
if "delivery_time_deviation_orig" not in df.columns:
    df = df.rename(columns={"delivery_time_deviation": "delivery_time_deviation_orig"})

df["lead_time_days_new"] = (df["actual_delivery_date"] - df["order_date"]).dt.days
df["expected_lead_time_days"] = (df["expected_delivery_date"] - df["order_date"]).dt.days
df["delivery_time_deviation"] = (df["actual_delivery_date"] - df["expected_delivery_date"]).dt.days  # + late, - early
df["on_time"] = (df["delivery_time_deviation"] <= 0).astype(int)

assert (df["order_date"] <= df["dispatch_date"]).all()
assert (df["dispatch_date"] <= df["actual_delivery_date"]).all()
assert df["lead_time_days_new"].gt(0).all()

needed = ["order_date", "dispatch_date", "expected_delivery_date", "actual_delivery_date",
          "lead_time_days_new", "delivery_time_deviation", "on_time"]
print("Missing columns:", [c for c in needed if c not in df.columns])
print(df[["lead_time_days_new", "expected_lead_time_days", "delivery_time_deviation"]].describe().T)
print("On-time rate:", round(df["on_time"].mean(), 3))
```

```python
rng = np.random.default_rng(SEED)          # reset so the full sequence reproduces
sup_quality = pd.Series(rng.beta(5, 2, df[SUP].nunique()), index=df[SUP].unique())
q = df[SUP].map(sup_quality).to_numpy()
ctry_transit = pd.Series(rng.integers(3, 26, df[CTRY].nunique()), index=df[CTRY].unique())

order_date = START + pd.to_timedelta(rng.integers(0, 730, n), unit="D")
planned_proc = rng.integers(1, 6, n)
planned_transit = np.clip(df[CTRY].map(ctry_transit).to_numpy() + rng.integers(-2, 3, n), 3, None)
expected_dispatch = order_date + pd.to_timedelta(planned_proc, unit="D")
expected_delivery = expected_dispatch + pd.to_timedelta(planned_transit, unit="D")

# Recalibrated: smaller delay rates, and a chance of on-schedule/early that grows with quality
dispatch_delay = rng.poisson((1 - q) * 1.5)
transit_delay = rng.poisson((1 - q) * 1.5)
early = rng.binomial(1, 0.4 * q) * rng.integers(1, 3, n)
actual_dispatch = expected_dispatch + pd.to_timedelta(dispatch_delay, unit="D")
actual_delivery = actual_dispatch + pd.to_timedelta(planned_transit + transit_delay - early, unit="D")

df["order_date"] = order_date
df["dispatch_date"] = actual_dispatch
df["expected_delivery_date"] = expected_delivery
df["actual_delivery_date"] = actual_delivery

df["lead_time_days_new"] = (df["actual_delivery_date"] - df["order_date"]).dt.days
df["expected_lead_time_days"] = (df["expected_delivery_date"] - df["order_date"]).dt.days
df["delivery_time_deviation"] = (df["actual_delivery_date"] - df["expected_delivery_date"]).dt.days
df["on_time"] = (df["delivery_time_deviation"] <= 0).astype(int)

assert (df["order_date"] <= df["dispatch_date"]).all()
assert (df["dispatch_date"] <= df["actual_delivery_date"]).all()
print(df[["lead_time_days_new", "expected_lead_time_days", "delivery_time_deviation"]].describe().T)
print("On-time rate:", round(df["on_time"].mean(), 3))
print("Shipments with lead time < 4 days:", (df["lead_time_days_new"] < 4).sum())
```

```python
g = df.groupby(SUP)
sup = pd.DataFrame({
    "avg_lead_time_days": g["lead_time_days_new"].mean(),
    "lead_time_std": g["lead_time_days_new"].std(),
    "avg_delivery_deviation": g["delivery_time_deviation"].mean(),
    "avg_late_days": g["delivery_time_deviation"].apply(lambda x: x.clip(lower=0).mean()),
    "on_time_rate": g["on_time"].mean(),
    "fulfilment_perf": g["fulfil_score"].mean(),
    "cargo_perf": g["cargo_score"].mean(),
    "n_shipments": g.size(),
})

late = sup["avg_late_days"]
sup["delay_score"] = 1 - (late - late.min()) / (late.max() - late.min())

WEIGHTS = {"on_time_rate": 0.30, "delay_score": 0.20, "fulfilment_perf": 0.25, "cargo_perf": 0.25}
sup["supplier_reliability_score_new"] = 100 * sum(sup[c] * w for c, w in WEIGHTS.items())

print(sup[["avg_lead_time_days", "lead_time_std", "on_time_rate",
           "fulfilment_perf", "cargo_perf", "supplier_reliability_score_new"]].describe().T)
```

```python
# Cell 3: sensitivity check, simulation check, merge back into df
from scipy.stats import spearmanr

# 1) Cargo direction sensitivity (mean of 1-x equals 1 minus mean of x)
flip = sup.copy()
flip["cargo_perf"] = 1 - flip["cargo_perf"]
sup_flip_score = 100 * sum(flip[c] * w for c, w in WEIGHTS.items())
print("Rank corr, cargo higher=better vs lower=better:",
      round(spearmanr(sup["supplier_reliability_score_new"], sup_flip_score)[0], 3))
print("Mean score if cargo flipped:", round(sup_flip_score.mean(), 1))

# 2) Simulation check: does the score recover the latent quality I generated?
# Only valid if sup_quality is still in memory from the date-generation cell
print("Rank corr with simulated quality:",
      round(spearmanr(sup["supplier_reliability_score_new"],
                      sup_quality.reindex(sup.index))[0], 3))

# 3) Merge supplier-level values back into df (drops old copies so re-running is safe)
add = sup[["avg_lead_time_days", "lead_time_std", "on_time_rate",
           "supplier_reliability_score_new", "n_shipments"]].add_prefix("sup_")
df = df.drop(columns=[c for c in add.columns if c in df.columns])
df = df.merge(add, left_on=SUP, right_index=True, how="left")
print(df.shape, df[add.columns].isna().sum().sum(), "missing after merge")
```

```python
# Cell 5: new target, leakage-free supplier history, time-based split
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import roc_auc_score, average_precision_score

df = df.sort_values("order_date").reset_index(drop=True)
df["late_2d"] = (df["delivery_time_deviation"] >= 2).astype(int)
print("Positive rate:", round(df["late_2d"].mean(), 3))

# Supplier history from EARLIER orders only (shift(1) excludes the current shipment)
g = df.groupby(SUP)
df["hist_n"] = g.cumcount()
df["hist_on_time"] = g["on_time"].transform(lambda s: s.shift(1).expanding().mean())
df["hist_late_rate"] = g["late_2d"].transform(lambda s: s.shift(1).expanding().mean())
df["hist_mean_dev"] = g["delivery_time_deviation"].transform(lambda s: s.shift(1).expanding().mean())

feat = ["hist_n", "hist_on_time", "hist_late_rate", "hist_mean_dev",
        "expected_lead_time_days", "route_risk_level", "weather_condition_severity",
        "customs_clearance_time", "warehouse_inventory_level", "shipping_costs"]
X = df[feat].apply(pd.to_numeric, errors="coerce")
y = df["late_2d"]

cutoff = df["order_date"].quantile(0.7)
tr, te = df["order_date"] < cutoff, df["order_date"] >= cutoff

m = HistGradientBoostingClassifier(random_state=42).fit(X[tr], y[tr])
p = m.predict_proba(X[te])[:, 1]
print("Test positive rate (baseline PR-AUC):", round(y[te].mean(), 3))
print("ROC-AUC:", round(roc_auc_score(y[te], p), 3))
print("PR-AUC:", round(average_precision_score(y[te], p), 3))
```

```python
# Cell 6: what is the model actually using?
from sklearn.inspection import permutation_importance

imp = permutation_importance(m, X[te], y[te], scoring="roc_auc",
                             n_repeats=5, random_state=42)
print(pd.Series(imp.importances_mean, index=feat).sort_values(ascending=False).round(4))

# Ablation: history-only vs real-columns-only
hist_cols = ["hist_n", "hist_on_time", "hist_late_rate", "hist_mean_dev"]
real_cols = ["route_risk_level", "weather_condition_severity", "customs_clearance_time",
             "warehouse_inventory_level", "shipping_costs", "expected_lead_time_days"]
for name, cols in [("history only", hist_cols), ("real columns only", real_cols)]:
    mm = HistGradientBoostingClassifier(random_state=42).fit(X.loc[tr, cols], y[tr])
    print(name, "ROC-AUC:", round(roc_auc_score(y[te], mm.predict_proba(X.loc[te, cols])[:, 1]), 3))
```
